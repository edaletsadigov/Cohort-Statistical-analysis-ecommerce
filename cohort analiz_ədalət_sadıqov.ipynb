## 3. Weekly retention analysis and final summary

### Weekly cohort

```python
orders_cleaned['order_week'] = (
    orders_cleaned['order_date']
    - pd.to_timedelta(orders_cleaned['order_date'].dt.weekday, unit='D')
).dt.normalize()

orders_cleaned['first_purchase_week'] = (
    orders_cleaned['first_order_date']
    - pd.to_timedelta(orders_cleaned['first_order_date'].dt.weekday, unit='D')
).dt.normalize()

orders_cleaned[['order_date', 'order_week', 'first_order_date', 'first_purchase_week']].head()
```

```python
num_of_weekly_cohorts = orders_cleaned['first_purchase_week'].nunique()

print(f'Total count of weekly cohort: {num_of_weekly_cohorts}')
```

- `order_week`is first day of order week for each customer
- `first_purchase_week` is first day of the week which customer purchased in that week

### Weekly cohort lifetime

```python
orders_cleaned['cohort_lifetime_weekly'] = (orders_cleaned['order_week'] - orders_cleaned['first_purchase_week']).dt.days // 7

orders_cleaned[['first_purchase_week', 'order_week', 'cohort_lifetime_weekly']].head()
```

```python
print(f'Minimum : {orders_cleaned['cohort_lifetime_weekly'].min()}')
print(f'Maximum :{orders_cleaned['cohort_lifetime_weekly'].max()}')
print(f'Negative count: {(orders_cleaned['cohort_lifetime_weekly']<0).sum()}')
```

###  Final Summary

**Key observations:**

1. **Data volume and cleaning.** Of 286,392 order lines, only 141,675 (~49.5%) carry a successful status (`complete`, `received`, `paid`, `closed`); the other half were canceled, refunded, or still in progress (`cod`, `pending`, etc.). This points to a serious operational issue (payment, delivery) worth separate investigation.

2. **Cohort sizes are highly uneven.** The Dec-2020 cohort (14,020 customers) is 3–9x larger than every other month, pointing to a seasonal campaign. Size dropped sharply afterward (Jan/Feb 2021), with smaller secondary waves in Apr and Jun 2021.

3. **Retention is extremely low.** Average retention across weekly cohorts: week 0 → 100%, week 1 → ~6.5%, week 2 → ~6.0%, week 3 → ~3.4%. About 94% of customers never come back after their first purchase — a one-time-buyer pattern dominates.

4. **ARPU behaves counterintuitively.** The small group that does return spends **more** on average than first-time buyers (ARPU rises with lifetime in most cohorts). The loyal segment is small but high-value — focusing on it could drive disproportionate revenue.

5. **Churn is essentially irreversible.** Cumulative churn stabilizes above 90% after week 1 — once lost, a customer almost never returns.

**Business recommendations:**

- **First-week horror.** The biggest retention drop happens between week 0 and week 1. Targeted email/SMS and a second-purchase discount coupon should be tested in this time.
- **Campaign-driven customers.** Large one-off campaign cohorts (like Dec-2020) tend to retain worse — a dedicated win-back campaign should target this segment.
- **Loyal segment.** Rising ARPU over lifetime shows the small returning group is high-value — retain it with a VIP program or loyalty points.
- **Operational quality.** Half of all orders being canceled/refunded is a standalone problem (payment failure, stock, delivery delay) that needs its own operational analysis.

## RFM Segmentation

```python
snapshot_date = orders_cleaned['order_date'].max() + pd.Timedelta(days=1)

rfm = orders_cleaned.groupby('cust_id').agg(
    last_purchase=('order_date', 'max'),
    frequency=('order_id', 'nunique'),
    monetary=('revenue', 'sum')
).reset_index()

rfm['recency'] = (snapshot_date - rfm['last_purchase']).dt.days

print(f'Snapshot Date: {snapshot_date.date()}')
rfm.head()
```

- `Recency` = Days since the last purchase
- `Frequency` = Number of unique orders
- `Monetary` = Total amount spent
- `Snapshot` Date = The day after the dataset’s last order date

```python
rfm['R_score'] = pd.qcut(rfm['recency'].rank(method='first'), 5, labels=[5, 4, 3, 2, 1]).astype(int)
rfm['F_score'] = pd.qcut(rfm['frequency'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5]).astype(int)
rfm['M_score'] = pd.qcut(rfm['monetary'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5]).astype(int)

rfm['RFM_score'] = rfm['R_score'].astype(str) + rfm['F_score'].astype(str) + rfm['M_score'].astype(str)
rfm.head()
```

Scores are based on the dataset’s quintiles

- `Recency:` lower days = better → score 5
- `Frequency / Monetary:` higher = better → score 5
- `.rank(method='first')` helps qcut handle duplicate values without errors.

```python
def assign_segment(row):
    r, f, m = row['R_score'], row['F_score'], row['M_score']
    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'
    elif r <= 2 and f >= 4 and m >= 4:
        return "Can't Lose Them"
    elif r >= 3 and f >= 3 and m >= 3:
        return 'Loyal Customers'
    elif r >= 4 and f <= 2:
        return 'New Customers'
    elif r <= 2 and f >= 3:
        return 'At Risk'
    elif r <= 2 and f <= 2 and m <= 2:
        return 'Hibernating/Lost'
    elif r >= 3 and f <= 3 and m <= 3:
        return 'Promising'
    else:
        return 'Potential Loyalists'

rfm['segment'] = rfm.apply(assign_segment, axis=1)
rfm['segment'].value_counts()
```

Customers are segmented based on their RFM scores:

- `Champions`: High R, F, and M
- `Can’t Lose Them`: Low R, but high F and M
- `Loyal Customers`: High overall RFM scores
- `New Customers`: Recent but low-frequency buyers
- `At Risk`: Low R, but relatively high F
- `Hibernating/Lost`: Low R, F, and M
- `Promising`: Recent customers with moderate F and M
- `Potential Loyalists`: Remaining customers with loyalty potential.

```python
segment_summary = rfm.groupby('segment').agg(
    customers=('cust_id', 'count'),
    revenue=('monetary', 'sum'),
    avg_frequency=('frequency', 'mean'),
    avg_recency=('recency', 'mean'),
).reset_index()

segment_summary['revenue_share_%'] = (segment_summary['revenue'] / segment_summary['revenue'].sum() * 100).round(1)
segment_summary['customer_share_%'] = (segment_summary['customers'] / segment_summary['customers'].sum() * 100).round(1)
segment_summary = segment_summary.sort_values('revenue', ascending=False).reset_index(drop=True)

segment_summary.round(1)
```

```python
fig_rfm = px.scatter(
    segment_summary,
    x='avg_recency', y='avg_frequency',
    size='customers', color='revenue_share_%',
    text='segment',
    size_max=80,
    color_continuous_scale='RdYlGn',
    title='RFM Segment Map — Recency vs Frequency (bubble size = customers, colour = revenue share)',
    labels={'avg_recency': 'Avg Recency (days since last order) — lower is better',
            'avg_frequency': 'Avg Frequency (orders per customer)',
            'revenue_share_%': 'Revenue share (%)'}
)
fig_rfm.update_traces(textposition='top center', textfont=dict(size=11, color='black'))
fig_rfm.update_xaxes(autorange='reversed')  # low recency (recent buyers) shown on the right
fig_rfm.update_layout(height=560, coloraxis_colorbar_title='Revenue share (%)')
fig_rfm.show()

top2 = segment_summary.sort_values('revenue', ascending=False).iloc[:2]
print(f"Champions + Can't Lose Them: {top2['customer_share_%'].sum():.1f}% of customers "
      f"generate {top2['revenue_share_%'].sum():.1f}% of revenue.")
```

`Revenue is highly concentrated`: Only `Champions` (12.4% of customers) generate 44.5% of total revenue. `Champions + Can't Lose Them` represent just 21.0% of customers, but generate `67.3%` of revenue — nearly two-thirds of revenue depends on a small customer group.

The `largest` group is not always the most valuable: `Loyal Customers` make up 20.0% of customers and contribute 20.5% of revenue. Meanwhile, `Promising` (18.0%) and `Hibernating/Lost` (16.0%) together represent over one-third of customers but generate only 1.9% of revenue.

- `total_revenue=('revenue','sum')` — Total revenue generated by each customer.
- `n_orders=('order_id','nunique')` — Counts unique orders, avoiding duplicate line-items within the same order.
- `region / gender` — Uses the first value because these attributes are constant per customer.
- `payment_pre`f — Uses .mode() to find the customer’s most common payment method; .iat[0] selects the first if multiple modes exist.

```python
cust_statistic['aov'] = (cust_statistic['total_revenue'] / cust_statistic['num_orders']).round(2)
cust_statistic['repeated'] = cust_statistic['num_orders'] > 1
cust_statistic.sample(5)
```

### Payment Method vs Repeat Purchase

```python
cust_statistic.columns.to_list()
```

```python
top_pay =  cust_statistic['payment_preference'].value_counts().head(5).index
sub = cust_statistic[cust_statistic['payment_preference'].isin(top_pay)]
ct = pd.crosstab(sub['payment_preference'],sub['repeated'])

ct
```

```python
chi2, p_val, dof, expected = stt.chi2_contingency(ct)
n = ct.values.sum()
cramers = np.sqrt(chi2/(n*(min(ct.shape)-1)))


print('H0: Payment method and repeat-purchase status are independent')
print('H1: Payment method and repeat-purchase status are associated')
print(ct)
print(f"\nChi2 = {chi2:.1f}, dof = {dof}, p-value = {p_val:.2e}, alpha = 0.05")
print(f"Cramér's V (for effectivity) = {cramers:.3f}")

print(f"İn conclusion: {'Reject H0' if p_val < 0.05 else 'Fail to reject H0'} — statistically significant, "
      f"effect size is {'small-to-moderate' if cramers < 0.3 else 'moderate-to-large'} (V={cramers:.2f}).")
```

A strong statistical association was found between payment method and repeat-purchase behavior `(Chi2=3325.3, p<0.001)`, with Cramér's V=0.29, indicating a moderate association.

Repeat-purchase rates vary significantly by payment method:

- `easypay_voucher: 62.9% (highest)`
- `Payaxis: 46.9%`
- `Easypay: 46.0%`
- `bankalfalah: 31.0%`
- `cod: 24.3% (lowest; largest group — 23,080 customers)`

Conclusion: cod users form the largest segment but have the lowest repeat rate, while `easypay_voucher` users show the `highest loyalty`. This is an association, not causation. Encouraging cod users to adopt alternative payment methods could be a potential retention strategy.

### Region vs AOV

```python
cust_statistic.columns.to_list()
```

```python
region_groups = [
    cust_statistic.loc[
        cust_statistic['region'] == region, 'aov'
    ].dropna()
    for region in cust_statistic['region'].dropna().unique()
]

statistic, p_value = stt.kruskal(*region_groups)
print("H0: AOV distribution is the same across all regions")
print("H1: at least one region's AOV distribution differs")
print(f"H-statistic = {statistic:.3f}, p-value = {p_value:.3f}, alpha = 0.05")
print(f"Conclusion: {'Reject H0' if p_value < 0.05 else 'Fail to reject H0'} — "
      f"no statistically significant difference in AOV across regions.")

print(f'H-statistic: {statistic:.3f}')
print(f'p-value: {p_value:.4g}')
```
